### The Term Structure of Interest Rates

> *Source: Lecture 02, `BFIN-WS2026-Lecture_02_bc.ipynb`, cell 25*


#### Spot rate
*   The yield to maturity on zero-coupon bonds.
*   The rate that prevails today for a time period corresponding to the zero's maturity.


#### Short rate
*   The one-period interest rate that prevails at a particular date. Future short rates are uncertain today.



#### Forward rate
*   The one-year rate implied by today's spot rates that makes an $n$-year zero-coupon investment equivalent to an $(n - 1)$-year zero-coupon investment rolled over for one year at that rate.
*   It equals the expected future short rate under the expectations hypothesis. A risk premium can make the two differ.
    *   If short-term investors dominate, the forward rate tends to exceed the expected future short rate to compensate for long-term bond price risk.
    *   If long-term investors dominate, the forward rate can be below the expected future short rate because rolling over short-term bonds involves reinvestment risk.


---

### The Term Structure of Interest Rates

> *Source: Lecture 02, `BFIN-WS2026-Lecture_02_bc.ipynb`, cell 26*

#### Liquidity
* Liquidity refers to the ability to sell an asset easily at a predictable price.
* Long-term bonds have greater price risk, considered less liquid, must offer a premium.
* Liquidity Premium
    * The liquidity premium compensates short-term investors for the uncertainty about the price at which they will be able to sell their long-term bonds at the end of the year.

#### The Expectations Hypothesis
*   The forward rate equals the market consensus expectation of the future short interest rate,
*   $f_n = E(r_n)$, and liquidity premiums are zero.
*   Suggests that the yields on long-term bonds depend only on expectations of future short rates.
    *   The forward rates derived from the yield curve can be used to infer market expectations of future short rates.\

#### Liquidity Preference Theory
*   Short-term investors will be unwilling to hold long-term bonds unless the forward rate exceeds the expected short interest rate.
*   Long-term investors will be unwilling to hold short bonds unless the expected short interest rate exceeds the forward rate.
*   The liquidity preference theory of the term structure argues that short-term investors dominate the market so that the forward rate will generally exceed the expected short rate.

$$ f_n = E(r_n) + \text{Liquidity Premium} $$

### Duration

> *Source: Lecture 03, `BFIN-WS2026-Lecture_03_bc.ipynb`, cell 6*

*   A measure of the effective maturity of a bond
*   The weighted average of the times until each payment is received
*   The weights are proportional to the present value of the payment
    *   Duration = Maturity for zero coupon bonds
    *   Duration < Maturity for coupon bonds

$$D = \sum_{t=1}^{T} t \times w_t \qquad \text{where} \qquad w_t = \frac{CF_t/((1+y)^t)}{P}$$

*   Duration is a key concept in fixed-income portfolio management.
    *   A simple summary statistic of the effective average maturity of the portfolio.
    *   An essential tool in immunizing portfolios from interest rate risk.
    *   A measure of the interest rate sensitivity of a portfolio.

*$CF_t$ is the cash flow at time t; $P$ is the price of the bond, and $y$ is the yield to maturity (YTM).*

---

### Duration-Price Relationship

> *Source: Lecture 03, `BFIN-WS2026-Lecture_03_bc.ipynb`, cell 7*

*   Price change is proportional to duration.
*   The percentage change in bond price is the product of modified duration and the change in the bond's yield to maturity.

$$\frac{\Delta P}{P} = -D \times \frac{\Delta y}{1+y}$$

$$\frac{\Delta P}{P} = -D_M \Delta y \qquad \text{where} \qquad D_M = \frac{D}{1 + y} \qquad \text{and} \qquad D_M \quad \text{is modified duration}$$

*   Rule 1: The duration of a zero-coupon bond equals its time to maturity.

*   Rule 2: Holding maturity constant, a bond's duration is higher when the coupon rate is lower.

*   Rule 3: Holding the coupon rate constant, a bond's duration generally increases with its time to maturity.

*   Rule 4: Holding other factors constant, the duration of a coupon bond is higher when the bond's yield to maturity is lower.

*   Rule 5: The duration of a level perpetuity is equal to: $\frac{1+y}{y}$

### Interest Rate Risk

> *Source: Lecture 03, `BFIN-WS2026-Lecture_03_bc.ipynb`, cell 12*


* Modified duration measures the sensitivity of a bond's price to a small change in its annual yield $y$.

$$
P = \sum_{k=1}^{mT}\frac{CF_k}{(1+y/m)^k},
\qquad D_M=-\frac{1}{P}\frac{dP}{dy},
\qquad \Delta P\approx-PD_M\Delta y
$$

Here $m$ is the number of coupon payments per year and $T$ is maturity in years.

* Money (dollar) duration is $PD_M=-dP/dy$. 
    * For a portfolio, money durations add when yields move together.

* DV01 is the value of a one-basis-point yield change: 
    *   $\text{DV01}\approx PD_M\times0.0001$. 
    *   It is the approximate price decrease for a one-basis-point rise in yield; portfolio DV01s also add.

### Interest Rate Risk: Convexity

> *Source: Lecture 03, `BFIN-WS2026-Lecture_03_bc.ipynb`, cell 15*


* Modified duration gives a straight-line approximation to the price change for a small change in annual YTM.
* Convexity measures the curvature of the price-yield relationship. For an option-free bond with positive convexity, the actual price is above the duration-only estimate for both a yield rise and a yield fall.

Let $Y$ be annual YTM, $f$ the number of payments per year, and $k$ the payment period. With $CF_k$ including principal in the last period,

$$
P(Y)=\sum_{k=1}^{fT}\frac{CF_k}{(1+Y/f)^k}
$$

$$
\text{Annual convexity}
=\frac{1}{P f^2(1+Y/f)^2}
\sum_{k=1}^{fT}k(k+1)\frac{CF_k}{(1+Y/f)^k}
$$

$$
\frac{\Delta P}{P}\approx-D_M\Delta Y
+\frac{1}{2}\text{Annual convexity}(\Delta Y)^2
$$

### Passive Management: Immunization

> *Source: Lecture 03, `BFIN-WS2026-Lecture_03_bc.ipynb`, cell 23*

*   Control interest rate risk
*   Widely used by pension funds, insurance companies, and banks
*   A natural mismatch between asset and liability maturity structures.
    *   Bank liabilities are primarily the deposits owed to customers, most of which are short-term and, consequently, have low duration.
    *   Bank assets by contrast are composed largely of outstanding commercial and consumer loans or mortgages, which have longer duration.
    *   What happens when interest rates rise unexpectedly?
    *   What about the risks pension funds are exposed to?
*   The interest rate exposure of assets and liabilities is matched in the portfolio.
    *   Match the duration of the assets and liabilities.
    *   Price risk and reinvestment rate risk approximately offset near the matching point.
        *   With equal present values and matched dollar durations, assets approximately fund the liability after a small, parallel yield shift.
    *   Asset and liability values do not stay matched for every yield movement.
    *   Rebalancing is needed as time passes and yields change.

### Passive Management: Immunization 

> *Source: Lecture 03, `BFIN-WS2026-Lecture_03_bc.ipynb`, cell 24*


*   Duration matching balances the difference between the accumulated value of the coupon payments (reinvestment rate risk) and the sale value of the bond (price risk).
    *   When interest rates fall, the coupons grow less than in the base case, but the higher value of the bond offsets this.
    *   When interest rates rise, the value of the bond falls, but the coupons approximately offset it because they are reinvested at the higher rate.

*   Coupon paying bond and single-payment obligation.
    *   For a small, parallel yield shift, equally funded assets and liabilities with matched durations change in value by approximately the same amount.
    *   For greater changes in the interest rate, the present value curves diverge.
    *   Convexity
    *   Rebalancing
        *   Interest rate changes cause mismatch.
        *   Asset durations will change with time.
        *   Without rebalancing, durations will become unmatched.

### Passive Management: Cash Flow Matching and Dedication

> *Source: Lecture 03, `BFIN-WS2026-Lecture_03_bc.ipynb`, cell 25*


* Cash-flow matching constructs a bond portfolio whose cash flows equal the liability payments on the same dates.
    * A single known liability can be matched with a zero-coupon bond having the same maturity and payoff.
* Dedication extends cash-flow matching to a series of liabilities using zero-coupon and/or coupon bonds.
* An exact match removes interest-rate risk from funding those liabilities and requires no rebalancing for rate changes.
* The match can fail if promised bond payments are not made, embedded options alter cash flows, or the liabilities change.
* Exact matching may be difficult or costly when suitable bonds are unavailable or illiquid.

### Risk and Risk Aversion: Deriving $U$

> *Source: Lecture 04, `BFIN-WS2026-Lecture_04_bc.ipynb`, cell 27*


The investor chooses a portfolio to maximize expected utility:

$$
E[u(r)]=\sum_s p(s)u(r(s)).
$$

Here $u(r)$ is utility from a realized return. For a risk-averse investor, $u'(r)>0$ and $u''(r)<0$: higher returns increase utility, but at a diminishing rate.

Write the uncertain return as its mean plus a random deviation:

$$
r=\mu+\varepsilon,\qquad \mu=E(r),\qquad E(\varepsilon)=0,\qquad E(\varepsilon^2)=\sigma^2.
$$

To express expected utility in terms of mean and variance, approximate $u(\mu+\varepsilon)$ around $\mu$ using a second-order Taylor expansion:

$$
u(r)\approx u(\mu)+u'(\mu)(r-\mu)+\frac{1}{2}u''(\mu)(r-\mu)^2.
$$

Take expectations. Since $E(r-\mu)=0$ and $E[(r-\mu)^2]=\sigma^2$:

$$
E[u(r)]\approx u(\mu)+\frac{1}{2}u''(\mu)\sigma^2.
$$

The certainty-equivalent return $c$ is the sure return that gives the same utility: $u(c)=E[u(r)]$. Using $u(c)\approx u(\mu)+u'(\mu)(c-\mu)$:

$$
c\approx\mu+\frac{1}{2}\frac{u''(\mu)}{u'(\mu)}\sigma^2
=E(r)-\frac{1}{2}A\sigma^2,
\qquad A=-\frac{u''(\mu)}{u'(\mu)}>0.
$$

The mean-variance score $U$ approximates this certainty-equivalent return: expected return minus a risk penalty. The $1/2$ comes from the Taylor expansion; treating $A$ as constant is a modeling assumption.

For example, if $E(r)=10\%$, $\sigma=15\%$, and $A=4$:

$$
c\approx 0.10-\frac{1}{2}(4)(0.15)^2=0.055=5.5\%.
$$

The investor is approximately indifferent between this risky portfolio and a sure return of 5.5%.

### Portfolio Risk

> *Source: Lecture 04, `BFIN-WS2026-Lecture_04_bc.ipynb`, cell 36*


*   When calculating the expected return of the portfolio, we take the weighted average of the expected returns of the assets.

*   We cannot use the same method for variance or standard deviation (i.e., simply taking a weighted average of individual asset variances is incorrect due to the effect of covariances).

*   **Covariance:**
    *   Measures the directional relationship between the returns of two assets.
    *   At the portfolio level, the sign and magnitude of the covariances between pairs of assets significantly impact the overall volatility (risk) of the portfolio.

*   For two assets, $i$ and $j$, when considering $N$ possible states of the world (scenarios), each with a probability $p_k$:

    $ \text{Cov}(r_i, r_j) = p_1 \cdot [r_{i,1} - E(r_i)] \cdot [r_{j,1} - E(r_j)] + p_2 \cdot [r_{i,2} - E(r_i)] \cdot [r_{j,2} - E(r_j)] + \dots + p_N \cdot [r_{i,N} - E(r_i)] \cdot [r_{j,N} - E(r_j)] $

    $ \text{Cov}(r_i, r_j) = \sum_{k=1}^{N} p_k \cdot [r_{i,k} - E(r_i)] \cdot [r_{j,k} - E(r_j)] $

    Where:
    *   $\text{Cov}(r_i, r_j)$ is the covariance between the returns of asset $i$ and asset $j$.
    *   $N$ is the number of possible states/scenarios and $p_k$ is the probability of state/scenario $k$ occurring.
    *   $r_{i,k}$ is the return of asset $i$ if state/scenario $k$ occurs and $E(r_i)$ is the expected return of asset $i$.
    *   $r_{j,k}$ is the return of asset $j$ if state/scenario $k$ occurs and $E(r_j)$ is the expected return of asset $j$.

### A Single-Factor Market

> *Source: Lecture 06, `BFIN-WS2026-Lecture_06_bc.ipynb`, cell 4*


*   Let $R_i=r_i-r_f$ and $R_M=r_M-r_f$ be excess returns, with a fixed risk-free rate $r_f$.

*   Regression equation:
    $$R_i(t) = \alpha_i + \beta_i R_M(t) + e_i(t)$$
*   Expected return-beta relationship:
    $$E(R_i) = \alpha_i + \beta_i E(R_M)$$
*   Variance = Systematic risk + Firm-specific risk:
    $$\sigma_i^2 = \beta_i^2 \sigma_M^2 + \sigma^2(e_i)$$
*   Covariance = Product of betas $\times$ Market risk:
    $$\text{Cov}(R_i, R_j) = \beta_i \beta_j \sigma_M^2$$
*   Correlation =
    $$\text{Corr}(R_i, R_j) = \frac{\beta_i \beta_j \sigma_M^2}{\sigma_i \sigma_j} = \left(\frac{\beta_i\sigma_M}{\sigma_i}\right)\left(\frac{\beta_j\sigma_M}{\sigma_j}\right)$$
    $$= \text{Corr}(R_i, R_M) \times \text{Corr}(R_j, R_M)$$

### Single-Index Model: Variance and Covariance

> *Source: Lecture 06, `BFIN-WS2026-Lecture_06_bc.ipynb`, cell 5*


Let $m=R_M-E(R_M)$, so $E(m)=0$ and $E(m^2)=\sigma_M^2$. 

Assume $E(e_k)=0$ and $\operatorname{Cov}(m,e_k)=0$ for each asset $k$, and $\operatorname{Cov}(e_i,e_j)=0$ for $i\ne j$. 

Thus $E(me_k)=0$ and $E(e_ie_j)=0$.

Subtract the expected return from the model:

$$
\begin{aligned}
R_k-E(R_k)
&=\alpha_k+\beta_kR_M+e_k-[\alpha_k+\beta_kE(R_M)]\\
&=\beta_k[R_M-E(R_M)]+e_k=\beta_km+e_k.
\end{aligned}
$$

Variance:

$$
\begin{aligned}
\sigma_i^2
&=E\!\left[(R_i-E(R_i))^2\right]\\
&=E\!\left[(\beta_i m+e_i)^2\right]\\
&=E\!\left[\beta_i^2m^2+2\beta_i me_i+e_i^2\right]\\
&=\beta_i^2E(m^2)+2\beta_iE(me_i)+E(e_i^2)\\
&=\beta_i^2\sigma_M^2+0+\sigma^2(e_i).
\end{aligned}
$$

Covariance, for $i\ne j$:

$$
\begin{aligned}
\operatorname{Cov}(R_i,R_j)
&=E\!\left[(R_i-E(R_i))(R_j-E(R_j))\right]\\
&=E\!\left[(\beta_i m+e_i)(\beta_j m+e_j)\right]\\
&=E\!\left[\beta_i\beta_jm^2+\beta_i me_j+\beta_j me_i+e_ie_j\right]\\
&=\beta_i\beta_jE(m^2)+\beta_iE(me_j)+\beta_jE(me_i)+E(e_ie_j)\\
&=\beta_i\beta_j\sigma_M^2+0+0+0.
\end{aligned}
$$

### Covariance estimation and portfolio choice

> *Source: Lecture 06, `BFIN-WS2026-Lecture_06_bc.ipynb`, cell 19*


* Mean-variance optimization requires a covariance matrix. 

    *   For $n$ stocks, estimating every variance and covariance directly requires $n(n+1)/2$ distinct entries. 

    *   Sampling error in these estimates can make optimal portfolio weights unstable.

*   The single-index model approximates cross-stock covariance as $\beta_i\beta_j\sigma_M^2$ and adds firm-specific variance to each stock's own variance. 
    *   This reduces the number of risk estimates, but assumes firm-specific shocks are uncorrelated.

*   The tradeoff is fewer noisy estimates against a more restrictive picture of how stocks move together.

### The Capital Asset Pricing Model (CAPM)

> *Source: Lecture 06, `BFIN-WS2026-Lecture_06_bc.ipynb`, cell 22*


* Portfolio theory identifies preferred portfolios given expected returns, variances, and covariances. 

* CAPM asks which expected returns are consistent with investors choosing portfolios and markets clearing.

* Under CAPM's common expectations and common risk-free rate, every investor combines the risk-free asset with the same tangency portfolio; risk aversion determines the mix. 
    *   Since aggregate risky holdings must equal market supply, this tangency portfolio is the value-weighted market portfolio of all traded risky assets. 

    * Under these assumptions, the market portfolio is mean-variance efficient.

---

### The Capital Asset Pricing Model (CAPM)

> *Source: Lecture 06, `BFIN-WS2026-Lecture_06_bc.ipynb`, cell 23*


**Assumptions**

* Investors share a planning horizon and expectations about returns, variances, and covariances, and choose mean-variance optimal portfolios.
* They can borrow or lend without limit at the risk-free rate $r_f$.
* Assets are tradable and divisible; markets are competitive, with no taxes, transaction costs, or short-sale restrictions.

**Main result**

For total returns $r_i$ and $r_M$:

$$E(r_i)=r_f+\beta_i[E(r_M)-r_f].$$

The expected risk premium of security $i$ is $\beta_i$ times the expected market risk premium. Using the earlier excess-return notation $R_i=r_i-r_f$ and $R_M=r_M-r_f$, this is $E(R_i)=\beta_i E(R_M)$.

**Beta**

$$\beta_i=\frac{\operatorname{Cov}(r_i,r_M)}{\operatorname{Var}(r_M)}=\frac{\sigma_i\rho_{i,M}}{\sigma_M}.$$

---

### CAPM: covariance and the risk premium

> *Source: Lecture 06, `BFIN-WS2026-Lecture_06_bc.ipynb`, cell 27*


* The market risk premium depends on market variance and the effective aggregate risk-aversion coefficient $\bar A$:
  $$E(R_M)=\bar A\sigma_M^2.$$

* Security $i$ contributes $w_i\operatorname{Cov}(R_i,R_M)$ to market variance, where $w_i$ is its market weight:
  $$\operatorname{Cov}(R_i,R_M)=\sum_j w_j\operatorname{Cov}(R_i,R_j).$$

* In CAPM equilibrium, the expected premium compensates this covariance risk at the market's rate:
  $$E(R_i)=\frac{E(R_M)}{\sigma_M^2}\operatorname{Cov}(R_i,R_M)=\beta_i E(R_M).$$

Since $R_i=r_i-r_f$, the expected total return is:
$$E(r_i)=r_f+\beta_i[E(r_M)-r_f].$$

---

### CAPM for portfolios

> *Source: Lecture 06, `BFIN-WS2026-Lecture_06_bc.ipynb`, cell 30*


For a fully invested portfolio, $\sum_i w_i=1$. Expected returns and betas are weighted averages:

$$E(r_P)=\sum_i w_i E(r_i),\qquad \beta_P=\sum_i w_i\beta_i.$$

If the component assets satisfy CAPM, so does the portfolio:

$$E(r_P)=\sum_i w_i\left[r_f+\beta_i(E(r_M)-r_f)\right]
=r_f+\beta_P[E(r_M)-r_f].$$

The market portfolio has beta one:

$$\beta_M=\frac{\operatorname{Cov}(r_M,r_M)}{\operatorname{Var}(r_M)}
=\frac{\sigma_M^2}{\sigma_M^2}=1.$$

---

### The SML and alpha

> *Source: Lecture 06, `BFIN-WS2026-Lecture_06_bc.ipynb`, cell 31*




* Alpha is the difference between a security's expected return and the return required by CAPM:
  $$E(r_i)-r_f=\alpha_i+\beta_i[E(r_M)-r_f].$$
* A security above the SML has positive alpha; one below it has negative alpha. CAPM equilibrium implies $\alpha_i=0$.

In the figure:
$$E(r_{\text{stock}})_{\text{CAPM}}=6\%+1.2(14\%-6\%)=15.6\%,
\qquad \alpha=17\%-15.6\%=1.4\text{ percentage points}.$$

### The Fama-French 3-Factor Model (FF3)

> *Source: Lecture 07, `BFIN-WS2026-Lecture_07_bc.ipynb`, cell 16*


APT does not identify the priced factors. Empirical models propose observable factors based on patterns in returns.

*   Empirical Motivation
    *   Fama and French (1993) added size and value factors to capture historical patterns in average returns that market beta did not fully explain.

*   The Regression Equation (Attribution)
    *   To measure risk exposures, we estimate the following time-series regression using excess returns ($R_{i,t} \equiv r_{i,t} - r_{f,t}$):
    $$
    R_{i,t} = \alpha_i + \beta_{i,M} R_{M,t} + \beta_{i,SMB} SMB_t + \beta_{i,HML} HML_t + e_{i,t}
    $$

*   The Pricing Equation (Equilibrium)
    *   If FF3 prices the asset correctly, the intercept $\alpha_i=0$, so its expected excess return is:
    $$
    E[R_i] = \beta_{i,M} \lambda_M + \beta_{i,SMB} \lambda_{SMB} + \beta_{i,HML} \lambda_{HML}
    $$
    *   Factor premiums are expected factor returns: $\lambda_M=E[R_M]$, $\lambda_{SMB}=E[SMB]$, and $\lambda_{HML}=E[HML]$.

---

### Fama-French Factors: Meaning and Construction

> *Source: Lecture 07, `BFIN-WS2026-Lecture_07_bc.ipynb`, cell 17*


| Factor | Construction | Interpretation of positive exposure |
| :--- | :--- | :--- |
| Market ($R_M=r_M-r_f$) | Broad market return minus the risk-free rate | Positive sensitivity to market excess returns |
| SMB (Small Minus Big) | Long small firms; short large firms | A tilt toward small-cap stocks |
| HML (High Minus Low) | Long high book-to-market firms; short low book-to-market firms | A tilt toward value stocks |

* In FF3, stocks are grouped by size and book-to-market into six portfolios (a $2\times3$ sort).
* SMB averages the small-minus-big spreads across value groups; HML averages the value-minus-growth spreads across size groups.
* These sorts help separate size and value effects, but do not guarantee zero exposure to other factors.

### The Carhart 4-Factor Model

> *Source: Lecture 07, `BFIN-WS2026-Lecture_07_bc.ipynb`, cell 19*


*   The Momentum Anomaly
    *   Jegadeesh and Titman (1993) documented that past winners tend to outperform past losers over horizons of 3 to 12 months.
    *   FF3 did not explain this momentum effect.

*   Augmenting the Model
    *   Carhart (1997) added momentum to FF3.
    *   WML (Winners Minus Losers): the return on a portfolio long past winners and short past losers.

*   The Regression Equation
    $$
    R_{i,t} = \alpha_i + \beta_{i,M} R_{M,t} + \beta_{i,SMB} SMB_t + \beta_{i,HML} HML_t + \beta_{i,WML} WML_t + e_{i,t}
    $$

*   Application
    *   A widely used benchmark for evaluating mutual fund performance after controlling for common factor exposures, including momentum.

### Smart Beta and Factor Investing

> *Source: Lecture 07, `BFIN-WS2026-Lecture_07_bc.ipynb`, cell 22*


Some apparent alpha can be explained by exposure to common return factors, such as size, value, and momentum.

Smart beta funds use transparent rules to select and/or weight securities by characteristics such as valuation ratios, profitability, or volatility.

| Implementation | Portfolio construction |
| :--- | :--- |
| Academic factors such as SMB and HML | Zero-net-investment long-short portfolios |
| Smart beta funds (often ETFs) | Typically long-only portfolios tilted toward selected characteristics |

For example, a typical value ETF overweights value stocks relative to a market-cap-weighted benchmark without shorting growth stocks.

> *Source: Lecture 07, `BFIN-WS2026-Lecture_07_bc.ipynb`, cell 23*

### Efficient Market Hypothesis (EMH) – Eugene Fama

> *Source: Lecture 08 PDF, slide 2*

* Background:
    * Knowledgeable investors constantly compete to find mispriced securities.
    * Many people are analyzing the same assets → No single investor has a lasting advantage.
* Information enters the market continuously and unpredictably.
    * News, announcements, and data arrive randomly → Future price movements difficult to forecast.
    * Even if there is some small predictability, it may be too tiny to exploit after costs (No free-lunch).
* Investors quickly act on new information to earn profits.
    * Their trading pushes prices up or down almost immediately ⇒ current prices reflect all available information.

**Given all currently available information, you cannot predict future price changes to earn abnormal profits.**

---

### Forms of Efficient Markets

> *Source: Lecture 08 PDF, slide 3*

| Form | Information | Implication |
|:--|:--|:--|
| Weak | Past information (historical prices & volume) | Technical analysis should not consistently outperform |
| Semi-strong | All public information (financial reports, news, ratios) | Fundamental analysis cannot generate abnormal returns |
| Strong | All — public, private/insider + past information | Even insiders cannot consistently earn abnormal returns |

### From EMH → Random Walk

> *Source: Lecture 08 PDF, slide 4*

* **Efficient Market Hypothesis:** Given all currently available information, you cannot predict future price changes to earn abnormal profits.
* **Random Walk Hypothesis:** Future price changes are unpredictable and independent of past changes.
$$E[P_{t+1}\mid I_t]=P_t$$
    * So $\Delta P$ behaves like a random walk.
* **From EMH → Random Walk:** If news is unpredictable, price movements become unpredictable too.

---

### Active vs Passive (Stock-picking vs index funds)

> *Source: Lecture 08 PDF, slide 5*

* Active management is expensive – analysts, research, travel → fees ~2% AUM + 20% incentive fees.
    * If markets are efficient, this cost is wasted, as any information they discover is already priced in.
* Index funds track the market (e.g., S&P 500, FTSE All-Share) and charge lower fees (0.1% or even 0%).
* Index investing has grown rapidly:
    * EU index fund share ↑ from 15% → 30% (2007–2017)
    * US index fund share ↑ from 19% → 43%
* But indexing cannot reach 100%.
    * If no one gathers information, prices won't adjust → markets become inefficient.
    * Some informed traders must exist for markets to stay efficient (Grossman-Stiglitz paradox).

---

### Is our market efficient?

> *Source: Lecture 08 PDF, slide 6*

* Efficiency depends on how quickly information moves into prices.
    * Few informed traders → fast adjustment → more efficient
    * Many noise traders → slow adjustment → less efficient
    * Efficiency is a spectrum: No fixed threshold — only relative comparison.
* Market efficiency is a heroic concept to defend:
    * Almost any evidence could contradict it. The serial correlation must be zero:
$$\rho_k=\text{Corr}(r_t,r_{t-k})=0\quad\forall k\ge 1$$
    * Eugene Fama studied 30 stocks in DJIA (1957–1962), he found: Weak form efficiency.

### Confirmation Bias and Hindsight Bias

> *Source: Lecture 08 PDF, slide 12*

| | Confirmation bias | Hindsight bias |
|:--|:--|:--|
| Definition | The tendency to seek, interpret, and remember information that confirms existing beliefs while ignoring contradictory evidence. | The inclination to see past events as predictable after they have occurred ("I knew it all along" effect). |
| Example | An investor bullish on tech stocks only reads positive news about the sector and disregards warnings about overvaluation. | After a market crash, investors claim they "knew it was coming," even though they didn't act beforehand. |
| Investment impact | Leads to overconfidence, poor diversification, and ignoring risks. | Distorts learning from past decisions, reinforces overconfidence, and may lead to repeated mistakes. |

---

### Representativeness Bias and Conservatism Bias

> *Source: Lecture 08 PDF, slide 13*

* **Representativeness bias**
    * Definition: Judging investments by stereotypes.
    * Example: "Big famous company → must be a good investment." (not always!)
* **Conservatism bias**
    * Definition: People resist updating beliefs even with new information.
    * Example: Holding a stock despite deteriorating earnings because past performance was strong.

---

### Mental Accounting Bias

> *Source: Lecture 08 PDF, slide 16*

* Definition
    * Mental accounting refers to the cognitive process where individuals categorize, evaluate, and keep track of financial activities by creating separate "mental accounts," even though money is fungible.
    * People treat money differently based on its source (salary vs. bonus) or intended use.
    * Leads to irrational financial decisions (e.g., splurging a tax refund while saving salary).
    * Violates the principle of fungibility in economics.
* Implications in Finance
    * Influences investment choices and risk-taking.
    * Can lead to suboptimal portfolio allocation.

---

### Overconfidence Bias

> *Source: Lecture 08 PDF, slide 17*

* Definition
    * Investors tend to overestimate their own knowledge, skill, and ability to pick winning stocks.
    * Even professionals are not immune.
* Investment impact
    * Overweight particular assets → portfolio becomes under-diversified
    * Trade too frequently → higher transaction costs & lower returns
* Example: More trading ≠ better performance
* Solution
    * Recognise overconfidence in themselves and clients
    * Encourage realistic expectations
    * Warn against excessive trading
    * Review past decisions honestly → What worked? What failed?

---

### Loss Aversion

> *Source: Lecture 08 PDF, slide 18 — Prospect Theory (Kahneman & Tversky, 1979)*

* Definition: People dislike losses more than they like equivalent gains.
* Investment impact
    * Hold losing stocks hoping they recover
    * Sell winning stocks too early → Disposition effect
    * Risk-seeking behaviour when losing
    * Tendency for inertia: do nothing when it may be valuable to act
* Example: Losing \$1 feels about twice as painful as winning \$1 feels good.
* S-shaped value function: steeper for losses than gains; convex for losses (risk-seeking); concave for gains (risk-averse).

---

### Self-Attribution Bias and Status Quo Bias

> *Source: Lecture 08 PDF, slide 19*

* **Self-attribution bias**
    * Investors credit success to skill but blame losses on bad luck → fail to learn from mistakes.
    * Investment impact: reinforces overconfidence; prevents meaningful long-term learning; may lead to repeated poor decisions.
* **Status quo bias**
    * Definition: Inertia: People avoid taking action, even if change is better.
    * Investment impact: delay in rebalancing portfolios; avoid switching funds despite better alternatives; fear of regret → procrastination.
    * Example: People rarely change default settings.

---

### Does Alpha Exist?

> *Source: Lecture 08 PDF, slide 20*

* What is alpha?
    * α = return above the market benchmark (risk-adjusted)
    * Jensen's alpha:
$$\alpha=r-\left[r_f+\beta\,(r_m-r_f)\right]$$
* EMH view: "Alpha should not exist"
    * Markets quickly incorporate all available information
    * Any temporary mispricing is arbitraged away instantly
    * Consistent alpha = extremely unlikely
    * If an investor beats the market → mostly luck
* Behavioural finance view: "Alpha exists because people are biased"
    * Human biases create predictable mispricing: skilled investors can earn alpha by exploiting these biases
    * Dalbar analysis: individual investors consistently achieve lower returns than the markets (S&P 500) due to behavioral biases like market timing and emotional decision-making.
